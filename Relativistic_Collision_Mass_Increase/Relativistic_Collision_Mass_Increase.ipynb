```python
import numpy as np
import matplotlib.pyplot as plt

# Constants
c = 3e8  # speed of light in m/s
m0 = 1.0  # rest mass (arbitrary units)

# Velocities: 0 to 0.99c
v = np.linspace(0, 0.99*c, 500)

# Lorentz factor
gamma = 1 / np.sqrt(1 - (v/c)**2)

# Final mass after collision
M_final = 2 * gamma * m0

# Plot
plt.figure(figsize=(8,5))
plt.plot(v/c, M_final, color='blue', lw=2)
plt.title("Relativistic Mass Increase in Collision")
plt.xlabel("Velocity (v/c)")
plt.ylabel("Final Rest Mass of Combined Object (units of m0)")
plt.grid(True)
plt.show()
```
